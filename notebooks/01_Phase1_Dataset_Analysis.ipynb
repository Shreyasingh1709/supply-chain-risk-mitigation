```python
categorical_columns = [
    "city_id",
    "store_id",
    "management_group_id",
    "first_category_id",
    "second_category_id",
    "third_category_id",
    "product_id",
    "holiday_flag",
    "activity_flag",
    "stockout_label"
]

numerical_columns = [
    "sale_amount",
    "discount",
    "precpt",
    "avg_temperature",
    "avg_humidity",
    "avg_wind_level"
]

print("Categorical features:", len(categorical_columns))
print("Numerical features:", len(numerical_columns))
print("Total features:", len(categorical_columns) + len(numerical_columns))
```

```python
from pyspark.ml import Pipeline
from pyspark.ml.feature import (
    StringIndexer,
    OneHotEncoder,
    VectorAssembler
)
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import (
    BinaryClassificationEvaluator,
    MulticlassClassificationEvaluator
)

print("All ML libraries imported successfully!")
```

```python
indexers = [
    StringIndexer(
        inputCol=c,
        outputCol=c + "_index",
        handleInvalid="keep"
    )
    for c in categorical_columns
]

encoder = OneHotEncoder(
    inputCols=[c + "_index" for c in categorical_columns],
    outputCols=[c + "_encoded" for c in categorical_columns]
)

print("Categorical encoding stages created!")
print("String indexers:", len(indexers))
print("One-hot encoder: Ready")
```

```python
encoded_columns = [
    c + "_encoded" for c in categorical_columns
]

assembler = VectorAssembler(
    inputCols=encoded_columns + numerical_columns,
    outputCol="features",
    handleInvalid="keep"
)

print("Feature assembler created successfully!")
print("Total input columns:", len(encoded_columns + numerical_columns))
print("Output column: features")
```

```python
preprocessing_pipeline = Pipeline(
    stages=indexers + [encoder, assembler]
)

print("Preprocessing pipeline created successfully!")
print("Total pipeline stages:", len(indexers) + 2)
```

```python
preprocessing_model = preprocessing_pipeline.fit(train_df)

print("Preprocessing model fitted successfully!")
```

```python
train_prepared = preprocessing_model.transform(train_df)

print("Training data transformed successfully!")

display(
    train_prepared.select(
        "features",
        "next_day_stockout_label"
    ).limit(5)
)
```

```python
test_prepared = preprocessing_model.transform(test_df)

print("Testing data transformed successfully!")

display(
    test_prepared.select(
        "features",
        "next_day_stockout_label"
    ).limit(5)
)
```

```python
from pyspark.ml.functions import vector_to_array
from pyspark.sql.functions import col, when, lit, sum as spark_sum

# Extract probability of stockout (class 1)
validation_scores = validation_predictions.withColumn(
    "stockout_probability",
    vector_to_array(col("probability"))[1]
)

# Test different thresholds
thresholds = [0.5, 0.45, 0.4, 0.35, 0.3]

for threshold in thresholds:

    results = validation_scores.withColumn(
        "tuned_prediction",
        when(
            col("stockout_probability") >= lit(threshold), 1
        ).otherwise(0)
    )

    metrics = results.agg(
        spark_sum(
            when(
                (col("next_day_stockout_label") == 1) &
                (col("tuned_prediction") == 1), 1
            ).otherwise(0)
        ).alias("TP"),

        spark_sum(
            when(
                (col("next_day_stockout_label") == 0) &
                (col("tuned_prediction") == 1), 1
            ).otherwise(0)
        ).alias("FP"),

        spark_sum(
            when(
                (col("next_day_stockout_label") == 1) &
                (col("tuned_prediction") == 0), 1
            ).otherwise(0)
        ).alias("FN")
    ).first()

    tp, fp, fn = metrics["TP"], metrics["FP"], metrics["FN"]

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0
    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    print(
        f"Threshold: {threshold} | "
        f"Precision: {precision:.4f} | "
        f"Recall: {recall:.4f} | "
        f"F1 Score: {f1:.4f}"
    )
```

```python
selected_threshold = 0.45

validation_final = validation_scores.withColumn(
    "tuned_prediction",
    when(
        col("stockout_probability") >= lit(selected_threshold),
        1
    ).otherwise(0)
)

print("Selected threshold:", selected_threshold)

validation_final.groupBy(
    "next_day_stockout_label",
    "tuned_prediction"
).count().orderBy(
    "next_day_stockout_label",
    "tuned_prediction"
).show()
```

```python
from pyspark.sql.functions import to_date, col

df_features = df_features.withColumn(
    "dt", to_date("dt")
)

# Training: March 28 – May 27
train_df = df_features.filter(
    col("dt") <= "2024-05-27"
)

# Validation: May 28 – June 10
validation_df = df_features.filter(
    (col("dt") >= "2024-05-28") &
    (col("dt") <= "2024-06-10")
)

# Final testing: June 11 – June 24
test_df = df_features.filter(
    (col("dt") >= "2024-06-11") &
    (col("dt") <= "2024-06-24")
)

print("Training records:", train_df.count())
print("Validation records:", validation_df.count())
print("Testing records:", test_df.count())
```

```python
from pyspark.ml import Pipeline
from pyspark.ml.feature import (
    StringIndexer,
    OneHotEncoder,
    VectorAssembler
)

categorical_columns = [
    "city_id", "store_id", "management_group_id",
    "first_category_id", "second_category_id",
    "third_category_id", "product_id",
    "holiday_flag", "activity_flag", "stockout_label"
]

numerical_columns = [
    "sale_amount", "discount", "precpt",
    "avg_temperature", "avg_humidity", "avg_wind_level"
]

indexers = [
    StringIndexer(
        inputCol=c,
        outputCol=c + "_index",
        handleInvalid="keep"
    )
    for c in categorical_columns
]

encoder = OneHotEncoder(
    inputCols=[c + "_index" for c in categorical_columns],
    outputCols=[c + "_encoded" for c in categorical_columns]
)

assembler = VectorAssembler(
    inputCols=[
        c + "_encoded" for c in categorical_columns
    ] + numerical_columns,
    outputCol="features"
)

preprocessing_pipeline = Pipeline(
    stages=indexers + [encoder, assembler]
)

preprocessing_model = preprocessing_pipeline.fit(train_df)

train_prepared = preprocessing_model.transform(train_df)
validation_prepared = preprocessing_model.transform(validation_df)
test_prepared = preprocessing_model.transform(test_df)

print("Preprocessing completed!")
print("Feature vector size:", len(
    test_prepared.select("features").first()["features"]
))
print("Test records:", test_prepared.count())
```
